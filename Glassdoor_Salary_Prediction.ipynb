## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Create a copy of original dataset
df1 = df.copy()

# Remove duplicate rows from dataset
df1.drop_duplicates(inplace=True)

# Remove unnecessary unnamed column if present
if 'Unnamed: 0' in df1.columns:
    df1.drop('Unnamed: 0', axis=1, inplace=True)

# Replace invalid values with NaN
df1.replace('-1', np.nan, inplace=True)
df1.replace(-1, np.nan, inplace=True)

# Remove rows where salary estimate is missing
df1 = df1[df1['Salary Estimate'].notnull()]

# Convert salary column to string type
salary = df1['Salary Estimate'].astype(str)

# Remove extra text from salary column
salary = salary.apply(lambda x: x.split('(')[0])

# Remove unwanted words and symbols
salary = salary.apply(lambda x: x.replace('Employer Provided Salary:', '')
                                   .replace('Per Hour:', '')
                                   .replace('$', '')
                                   .replace('K', '')
                                   .replace('per hour', '')
                                   .replace('Per Hour', '')
                                   .strip())

# Create minimum salary column
df1['min_salary'] = salary.apply(
    lambda x: int(x.split('-')[0].strip())
)

# Create maximum salary column
df1['max_salary'] = salary.apply(
    lambda x: int(x.split('-')[1].strip()) if '-' in x else int(x.strip())
)

# Create average salary column
df1['avg_salary'] = (df1['min_salary'] + df1['max_salary']) / 2

# Remove ratings attached with company names
df1['Company Name'] = df1['Company Name'].apply(
    lambda x: x.split('\n')[0]
)

# Extract state from location column
df1['job_state'] = df1['Location'].apply(
    lambda x: x.split(',')[-1].strip()
)

# Calculate company age using founded year
df1['company_age'] = df1['Founded'].apply(
    lambda x: 2026 - x if x > 0 else np.nan
)

# Calculate length of job description
df1['description_length'] = df1['Job Description'].apply(len)

# Extract important skills from job description
df1['Python'] = df1['Job Description'].apply(
    lambda x: 1 if 'python' in x.lower() else 0
)

df1['SQL'] = df1['Job Description'].apply(
    lambda x: 1 if 'sql' in x.lower() else 0
)

df1['Excel'] = df1['Job Description'].apply(
    lambda x: 1 if 'excel' in x.lower() else 0
)

df1['AWS'] = df1['Job Description'].apply(
    lambda x: 1 if 'aws' in x.lower() else 0
)

df1['Spark'] = df1['Job Description'].apply(
    lambda x: 1 if 'spark' in x.lower() else 0
)

# Simplify job titles into categories
def simplify_job(title):

    title = title.lower()

    if 'data scientist' in title:
        return 'Data Scientist'

    elif 'data engineer' in title:
        return 'Data Engineer'

    elif 'analyst' in title:
        return 'Data Analyst'

    elif 'machine learning' in title:
        return 'ML Engineer'

    elif 'manager' in title:
        return 'Manager'

    else:
        return 'Other'

# Apply function on job title column
df1['job_simplified'] = df1['Job Title'].apply(simplify_job)

# Display final dataset shape
print("Final Dataset Shape :", df1.shape)

# Display first 5 rows
df1.head()
````

### What all manipulations have you done and insights you found?

The dataset was cleaned and transformed to make it ready for analysis and machine learning. Duplicate records and unnecessary columns were removed, and invalid values were handled properly. The Salary Estimate column was cleaned and converted into min_salary, max_salary, and avg_salary columns. New features such as company_age, job_state, and description_length were created. Important skills like Python, SQL, Excel, AWS, and Spark were extracted from job descriptions, and job titles were simplified into categories. These manipulations helped in improving data quality and preparing the dataset for visualization, business analysis, and salary prediction modeling.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Statement: Jobs requiring Python skills offer higher average salaries compared to jobs that do not require Python skills.

Null Hypothesis (H₀): There is no significant difference in average salaries between jobs requiring Python skills and jobs not requiring Python skills.
Alternate Hypothesis (H₁): Jobs requiring Python skills have significantly higher average salaries than jobs not requiring Python skills.

#### 2. Perform an appropriate statistical test.

````python
# Import statistical testing library
from scipy.stats import ttest_ind

# Separate salary groups based on Python skill requirement
python_required = df1[df1['Python'] == 1]['avg_salary']

python_not_required = df1[df1['Python'] == 0]['avg_salary']

# Perform Independent T-Test
t_statistic, p_value = ttest_ind(python_required,
                                 python_not_required,
                                 nan_policy='omit')

# Display results
print("T-Statistic :", t_statistic)
print("P-Value :", p_value)
````

##### Which statistical test have you done to obtain P-Value?

An Independent T-Test (Two Sample T-Test) was performed to obtain the P-value. This statistical test is used to compare the average salaries of two independent groups: jobs requiring Python skills and jobs not requiring Python skills. The test helps determine whether the difference in average salaries between the two groups is statistically significant or occurred due to random chance.

##### Why did you choose the specific statistical test?

The Independent T-Test was chosen because the analysis involves comparing the mean average salaries of two independent groups: jobs that require Python skills and jobs that do not require Python skills. The target variable, avg_salary, is numerical, while the grouping variable, Python, contains two categories (0 and 1). The Independent T-Test is specifically designed to determine whether the difference between the means of two independent groups is statistically significant. Therefore, it is the most suitable statistical test for evaluating whether Python skills have a meaningful impact on salary levels in the technology industry.

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Statement: Larger companies tend to offer higher salaries compared to smaller companies in the technology industry.

Null Hypothesis (H₀): There is no significant difference in average salaries between companies of different sizes.
Alternate Hypothesis (H₁): There is a significant difference in average salaries between companies of different sizes.

#### 2. Perform an appropriate statistical test.

````python
# Import statistical testing library
from scipy.stats import f_oneway

# Remove missing values
company_salary_df = df1[['Size', 'avg_salary']].dropna()

# Create salary groups based on company size
salary_groups = []

for size in company_salary_df['Size'].unique():
    group = company_salary_df[company_salary_df['Size'] == size]['avg_salary']
    salary_groups.append(group)

# Perform One-Way ANOVA Test
f_statistic, p_value = f_oneway(*salary_groups)

# Display results
print("F-Statistic :", f_statistic)
print("P-Value :", p_value)
````

### 2. Handling Outliers

````python
# Select numerical columns
numerical_columns = ['avg_salary', 'min_salary', 'max_salary',
                     'company_age', 'description_length']

# Create boxplots to detect outliers
plt.figure(figsize=(15,8))

for i, column in enumerate(numerical_columns):

    plt.subplot(2,3,i+1)

    sns.boxplot(y=df1[column])

    plt.title(f'Boxplot of {column}')

plt.tight_layout()
plt.show()

# Outlier Treatment using IQR Method

for column in numerical_columns:

    Q1 = df1[column].quantile(0.25)
    Q3 = df1[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Remove outliers
    df1 = df1[(df1[column] >= lower_bound) &
              (df1[column] <= upper_bound)]

# Display dataset shape after outlier treatment
print("Dataset Shape After Outlier Removal :", df1.shape)
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

In this project, the Interquartile Range (IQR) method was used for outlier detection and treatment. First, boxplots were created to visually identify extreme values in numerical features such as average salary, minimum salary, maximum salary, company age, and description length. After identifying outliers, the IQR technique was applied by calculating the first quartile (Q1), third quartile (Q3), and the interquartile range (IQR = Q3 − Q1). Values lying below Q1 − 1.5 × IQR and above Q3 + 1.5 × IQR were treated as outliers and removed from the dataset.

The IQR method was chosen because it is simple, effective, and less affected by skewed data and extreme values. It helps improve data quality by removing abnormal observations that could negatively impact visualization results and machine learning model performance. This technique also helps create a more stable and reliable predictive model by reducing noise and inconsistencies in the dataset.

#### 4. Removing URLs & Removing words and digits contain digits.

````python
# Remove URLs & Remove words and digits containing digits

# Import regular expression library
import re

# Function to remove URLs
def remove_urls(text):
    return re.sub(r'http\S+|www\S+|https\S+', '', text)

# Function to remove words containing digits
def remove_words_digits(text):
    return re.sub(r'\w*\d\w*', '', text)

# Apply URL removal
df1['Job Description'] = df1['Job Description'].apply(remove_urls)

# Apply digit-word removal
df1['Job Description'] = df1['Job Description'].apply(remove_words_digits)

# Display first 5 rows
df1['Job Description'].head()
````

#### 5. Removing Stopwords & Removing White spaces

````python
# Remove Stopwords

# Import stopwords
from nltk.corpus import stopwords

# Download stopwords
import nltk
nltk.download('stopwords')

# Store English stopwords
stop_words = set(stopwords.words('english'))

# Function to remove stopwords
def remove_stopwords(text):
    words = text.split()
    filtered_words = [word for word in words if word not in stop_words]
    return " ".join(filtered_words)

# Apply function on Job Description column
df1['Job Description'] = df1['Job Description'].apply(remove_stopwords)

# Display first 5 rows
df1['Job Description'].head()
````

````python
# Remove White Spaces

# Function to remove extra white spaces
def remove_whitespace(text):
    return " ".join(text.split())

# Apply function on Job Description column
df1['Job Description'] = df1['Job Description'].apply(remove_whitespace)

# Display first 5 rows
df1['Job Description'].head()
````

#### 7. Tokenization

````python
# Tokenization

# Import nltk
import nltk

# Download required tokenizer resources
nltk.download('punkt')
nltk.download('punkt_tab')

# Import word tokenizer
from nltk.tokenize import word_tokenize

# Function for tokenization
def tokenize_text(text):
    return word_tokenize(str(text))

# Apply tokenization on Job Description column
df1['Tokenized_Text'] = df1['Job Description'].apply(tokenize_text)

# Display first 5 rows
df1[['Tokenized_Text']].head()
````

#### 8. Text Normalization

````python
# Normalizing Text (Lemmatization)

# Import nltk
import nltk

# Download WordNet resources
nltk.download('wordnet')
nltk.download('omw-1.4')

# Import WordNet Lemmatizer
from nltk.stem import WordNetLemmatizer

# Create lemmatizer object
lemmatizer = WordNetLemmatizer()

# Function for lemmatization
def lemmatize_text(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

# Apply lemmatization on tokenized text
df1['Lemmatized_Text'] = df1['Tokenized_Text'].apply(lemmatize_text)

# Display first 5 rows
df1[['Lemmatized_Text']].head()
````

##### Which text normalization technique have you used and why?

In this project, Lemmatization was used as the text normalization technique. Lemmatization converts words into their meaningful root or base form while preserving the actual meaning of the word. For example, words like “running”, “runs”, and “ran” are converted into the base word “run”. This technique was chosen because it produces more meaningful and grammatically correct words compared to stemming, which may sometimes generate incomplete or incorrect root words. Lemmatization helps improve text quality, reduces unnecessary word variations, and makes text data more consistent for further natural language processing and machine learning tasks.

#### 9. Part of speech tagging

````python
# POS Tagging

# Import nltk
import nltk

# Download required resources
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

# Import POS tagger
from nltk import pos_tag

# Function for POS tagging
def pos_tagging(tokens):
    return pos_tag(tokens)

# Apply POS tagging on tokenized text
df1['POS_Tagged_Text'] = df1['Tokenized_Text'].apply(pos_tagging)

# Display first 5 rows
df1[['POS_Tagged_Text']].head()
````

#### 10. Text Vectorization

````python
# Vectorizing Text

# Import TF-IDF Vectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Convert lemmatized tokens back into sentence format
df1['Final_Text'] = df1['Lemmatized_Text'].apply(lambda x: " ".join(x))

# Create TF-IDF vectorizer object
tfidf = TfidfVectorizer(max_features=500)

# Convert text data into numerical vectors
tfidf_matrix = tfidf.fit_transform(df1['Final_Text'])

# Convert TF-IDF matrix into DataFrame
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

# Display first 5 rows of vectorized text
tfidf_df.head()
````

##### Which text vectorization technique have you used and why?

In this project, the TF-IDF (Term Frequency–Inverse Document Frequency) vectorization technique was used to convert textual job descriptions into numerical form. TF-IDF assigns importance scores to words based on how frequently they appear in a document and how unique they are across the entire dataset. Words that occur frequently in a specific document but less frequently in other documents receive higher importance scores.

TF-IDF was chosen because it helps identify meaningful and important words while reducing the influence of very common words that do not contribute much to analysis. It is widely used in Natural Language Processing because it creates efficient numerical representations of text data for machine learning models. This technique improves text analysis quality and helps the model better understand important skills, technologies, and keywords present in job descriptions.
